```python
import pandas as pd
```

```python
raw_data = {'regiment': ['Nighthawks', 'Nighthawks', 'Nighthawks', 'Nighthawks', 'Dragoons', 'Dragoons', 'Dragoons', 'Dragoons', 'Scouts', 'Scouts', 'Scouts', 'Scouts'], 
        'company': ['1st', '1st', '2nd', '2nd', '1st', '1st', '2nd', '2nd','1st', '1st', '2nd', '2nd'], 
        'name': ['Miller', 'Jacobson', 'Ali', 'Milner', 'Cooze', 'Jacon', 'Ryaner', 'Sone', 'Sloan', 'Piger', 'Riani', 'Ali'], 
        'preTestScore': [4, 24, 31, 2, 3, 4, 24, 31, 2, 3, 2, 3],
        'postTestScore': [25, 94, 57, 62, 70, 25, 94, 57, 62, 70, 62, 70]}
```

```python
dataset = pd.DataFrame(raw_data, columns=['regiment', 'company', 'name', 'preTestScore', 'postTestScore'])
```

```python
dataset
```

# What is the mean preTestScore from the regiment Nighthawks?

```python
mean_nighthawks = dataset.groupby('regiment')['preTestScore'].mean()
print(mean_nighthawks)
print(mean_nighthawks.to_dict())
print(f"Mean preTestScore of Nighthawks", mean_nighthawks['Nighthawks'])
```

```python
mean_nighthawks = dataset.groupby('regiment')[['preTestScore']].mean()  # 2D Dataframe
mean_nighthawks
```

# What is the mean of each company's preTestScore?

# Present the mean preTestScores grouped by regiment and company

```python
reg_and_comp = dataset.groupby(['regiment', 'company'])['preTestScore'].mean().unstack()
reg_and_comp.ndim
```

```python
reg_and_comp = dataset.groupby(['regiment', 'company'])['preTestScore'].mean()
print(reg_and_comp.ndim)
print(reg_and_comp)
```

# Iterate over a group and print the name and the whole data from the regiment

```python
for name, group in dataset.groupby('regiment'):
    print("Regiment:", name)
    print(group)
    print("-" * 40)
```

```python
for (regiment, company), group in dataset.groupby(['regiment', 'company']):
    print("Regiment:", regiment, "| Company:", company)
    print(group)
    print("-" * 40)
```
