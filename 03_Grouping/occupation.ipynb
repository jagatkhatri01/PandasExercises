```python
import pandas as pd
```

```python
import numpy as np
```

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user"
dataset = pd.read_csv(url, sep='|')
```

```python
dataset.head(3)
```

```python
dataset.shape
```

# Discover what is the mean age per occupation

```python
dataset.groupby('occupation')['age'].mean().round(2)
```

# Discover the Male ratio per occupation and sort it from the most to the least

```python
gender_count = dataset.groupby('occupation')['gender'].value_counts().unstack()
gender_count
```

```python
gender_count['male_ratio'] = gender_count['M'] / (gender_count['M']+gender_count['F'])
gender_count.head(3)
```

```python
gender_count['male_ratio'].sort_values(ascending=False)
```

# For each occupation, calculate the minimum and maximum ages

```python
dataset.groupby('occupation')['age'].agg(['min', 'max'])
```

#  For each combination of occupation and gender, calculate the mean age

```python
dataset.groupby(['occupation', 'gender'])['age'].mean()
```

# For each occupation present the percentage of women and men

```python
gender_percentage = dataset.groupby('occupation')['gender'].value_counts(normalize=True).mul(100).round(2)
gender_percentage.head(5)
```
